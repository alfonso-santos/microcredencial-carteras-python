## 🧪 Ejercicio: Validación empírica del CAPM en el S&P 500

Queremos comprobar si el **modelo de mercado (CAPM)** se cumple empíricamente para los activos del **S&P 500**.

### 🔢 Datos

- Utiliza **datos diarios** desde **1995-01-01** hasta la fecha actual.
- Los tickers del S&P 500 están disponibles en el siguiente fichero CSV:



filepath = 'https://raw.githubusercontent.com/alfonso-santos/microcredencial-carteras-python-2023/main/Tema_5_APT/data/sp500_tickers.csv'


- Utiliza el índice **S&P 500 (`^GSPC`)** como proxy del mercado.

---

### ✅ Instrucciones

#### 1. Calcula para cada activo:
- Su **beta** respecto al índice S&P 500 usando una regresión sobre los rendimientos diarios.
- Su **rentabilidad media diaria**.

#### 2. Estima una regresión:
- Usa como variable dependiente las **rentabilidades medias** y como explicativa las **betas**.
- Representa gráficamente la relación y comenta si se ajusta a lo esperado por el CAPM.

#### 3. Conclusiones:
- ¿Qué conclusiones puedes extraer de la relación entre betas y rentabilidades medias?
- ¿Qué implicaciones tiene para la validez empírica del modelo?

#### 4. Analiza en distintos regímenes de mercado:
- Descarga los precios del S&P 500 y **grafícalos**.
- Elige manualmente un **periodo claramente alcista** (subida sostenida > 1 año) y un **periodo claramente bajista** (caída sostenida > 1 año).
- Repite los pasos 1 y 2 usando únicamente:
  - Datos del periodo alcista.
  - Datos del periodo bajista.

#### 4. Conclusiones:
- ¿Se ajusta mejor el CAPM en mercados alcistas y bajistas?
- ¿Cómo se comportan las betas y los retornos en cada periodo?
- ¿Qué implicaciones tienen los resultados para la validez empírica del modelo?

---

### 📌 Entregable

Un notebook que incluya:
- Descarga y preparación de los datos.
- Cálculo de rendimientos, betas y rentabilidades medias.
- Gráficas con la regresión estimada.
- Análisis comparativo de los distintos periodos y tus conclusiones razonadas.
