## 🧪 Exercise: Empirical Validation of the CAPM in the S&P 500

We want to examine whether the **market model (CAPM)** holds empirically for **S&P 500** stocks.

### 🔢 Data

- Use **daily data** from **1995-01-01** to the present.
- The S&P 500 tickers are available in the following CSV file:


filepath = 'https://raw.githubusercontent.com/alfonso-santos/microcredencial-carteras-python-2023/main/Tema_5_APT/data/sp500_tickers.csv'

- Use the **S&P 500 index (`^GSPC`)** as a proxy for the market.

---

### ✅ Instructions

#### 1. Calculate for each stock:
- Its **beta** relative to the S&P 500 index using a regression on daily returns.
- Its **average daily return**.

#### 2. Estimate a regression:
- Use **average returns** as the dependent variable and **betas** as the explanatory variable.
- Plot the relationship and discuss whether it is consistent with CAPM predictions.

#### 3. Conclusions:
- What conclusions can you draw from the relationship between betas and average returns?
- What are the implications for the empirical validity of the model?

#### 4. Analyze different market regimes:
- Download and **plot** S&P 500 prices.
- Manually select a **clearly bullish period** (sustained increase for more than one year) and a **clearly bearish period** (sustained decline for more than one year).
- Repeat steps 1 and 2 using only:
  - Data from the bullish period.
  - Data from the bearish period.

#### 4. Conclusions:
- Does the CAPM fit better in bullish or bearish markets?
- How do betas and returns behave in each period?
- What are the implications of the results for the empirical validity of the model?

### 📌 Deliverable

A notebook that includes:
