The **Python Applied to Portfolio Management** course is an academic program designed to combine financial theory with practical implementation using computational tools.

Portfolio management is a key area in finance, and a solid understanding of it is highly valued in the job market.

In this course, we will not only cover the fundamental pillars of portfolio theory, such as:

- The Markowitz efficient frontier

- The CAPM, APT, and Black-Litterman models

but we will also learn how to implement and validate these concepts using Python, one of the most versatile and widely used programming languages in finance.

The **specific objectives** of the course are:

* **Theoretical Foundations:** Develop a deeper understanding of modern portfolio theory and the key concepts underlying investment management.

* **Practical Application with Python:** Use Python to simulate, optimize, and validate investment strategies based on the theoretical concepts covered in the course.
