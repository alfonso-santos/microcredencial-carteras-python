## 1. Download data from Yahoo Finance

`yfinance` is a Python library that provides a simple interface for accessing and downloading financial data from Yahoo Finance. It makes it easy to obtain historical prices and information on stocks, currencies, bonds, indices, and other financial assets.

The library integrates naturally with `pandas`, since downloaded data are returned as a DataFrame.

You can find more information about the library here: [yfinance](https://pypi.org/project/yfinance/)

Install the library if necessary:

```python
pip install yfinance
```

````python
import yfinance as yf
````

````python
# Download Apple ('AAPL') data between 2000 and 2010

data_yahoo = yf.download(
    "AAPL",
    start="2000-01-01",
    end="2011-01-01",  # 'start' specifies the initial date (included)
    auto_adjust=False,  # False returns 'Close' and 'Adj Close' separately. True makes Close = Adj Close
    multi_level_index=False,  # We do not want a MultiIndex in this case
)  # 'end' specifies the final date (excluded)
````

The variable `data_yahoo` is a Pandas DataFrame containing the following information:

* **Date** (DataFrame index): date corresponding to each observation
* **Open**: opening price of the asset
* **High**: highest price reached during the period
* **Low**: lowest price reached during the period
* **Close**: closing price of the asset
* **Adj Close**: closing price adjusted for corporate events such as dividends and stock splits
* **Volume**: trading volume

Let's take a look at the DataFrame:

````python
data_yahoo
````

````python
data_yahoo.shape  # 2767 rows, 6 columns
````

We will usually work with closing prices. In particular, we are going to select the adjusted closing price column, `Adj Close`.

**NOTE:** Be careful when selecting data from a DataFrame. Depending on how the selection is made, the result may be either a Pandas Series or a DataFrame.

````python
# Selecting a single column returns a Series
serie_yahoo = data_yahoo.loc[:, "Adj Close"]

print(serie_yahoo)
print(type(serie_yahoo))
````

````python
# Selecting a list of columns (even if it contains only one column)
# returns a DataFrame
df1_yahoo = data_yahoo.loc[:, ["Adj Close"]]

print(df1_yahoo)
print(type(df1_yahoo))
````

In the next cell, we download data for several financial assets at the same time. We simply pass a list of tickers to `yf.download` instead of a single ticker:

````python
activos = ["AAPL", "MSFT", "AMZN", "GOOG", "TSLA", "NVDA", "ADBE", "NFLX"]
df_yahoo = yf.download(
    activos,
    start="2000-01-01",
    end="2011-01-01",
    auto_adjust=False,
    multi_level_index=True,
)
````

When we download several assets, `yfinance` returns a DataFrame with **MultiIndex columns** of the form `(attribute, asset)`.

The first level contains the attributes (`Open`, `High`, `Low`, `Close`, etc.) and the second level identifies the corresponding asset (`AAPL`, `MSFT`, etc.).

````python
df_yahoo
````

Let's practice the three most common selection cases:

2. Display the `Close` prices for `MSFT`.

````python
# 1. Display the `Open` prices for all assets.

# SOLUTION
solucion_1 = df_yahoo.loc[:, "Open"]
solucion_1.head()
````

````python
# 2. Display the `Close` prices for 'MSFT'.

# SOLUTION
idx = pd.IndexSlice
solucion_2 = df_yahoo.loc[:, idx["Close", "MSFT"]]
solucion_2.head()
````

````python
# 3. Display all available attributes for 'NVDA'.

# SOLUTION
idx = pd.IndexSlice
solucion_3 = df_yahoo.loc[:, idx[:, "NVDA"]]
solucion_3.columns = solucion_3.columns.droplevel(
    1
)  # Try commenting out this line. What happens?
solucion_3.head()
````
