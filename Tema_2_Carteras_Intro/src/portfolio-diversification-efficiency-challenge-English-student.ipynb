## 1. The Effect of Diversification on Portfolio Risk

We have seen that the variance of a two-asset portfolio is given by:

$$\sigma^2_p= \underset{\text{Asset 1}} {\underbrace{w_1^2\sigma^2_1}}+\underset{\text{Asset 2}} {\underbrace{w_2^2\sigma^2_2}}+ \underset{\text{Relationship between assets}} {\underbrace{2w_1w_2 \sigma_{12}}}$$

We can generalize this expression to a portfolio with $n$ assets:

$$\sigma^2_p = \underset{\text{Individual asset risk}} {\underbrace{\sum_{i=1}^n w_i^2 \sigma_i^2}} + \underset{\text{Relationships between assets}} {\underbrace{ \sum_{i=1}^n \sum_{j=1}^n w_i w_j \sigma_{ij}}} \quad \forall i \neq j$$

For an equally weighted portfolio, each asset has weight $w_i=\dfrac{1}{n}$. The portfolio variance can then be written as:

$$\sigma_p^2= \dfrac{1}{n^2}\sum_{i=1}^{n} \sigma _i^2 + \dfrac{1}{n^2}\sum_{i=1}^{n}\sum_{j=1}^{n}\sigma_{ij} \quad \forall i \neq j$$

Define:

* **Average individual asset variance**

$$\bar{V} = \dfrac{1}{n}\sum_{i=1}^{n} \sigma _i^2 \quad \Rightarrow \quad \sum_{i=1}^{n} \sigma _i^2 = n \bar{V}$$

* **Average covariance between distinct assets**

$$\bar{\sigma_{ij}} = \dfrac{1}{n(n-1)}\sum_{i=1}^{n}\sum_{j=1}^{n}\sigma_{ij} \quad \Rightarrow \quad  \sum_{i=1}^{n}\sum_{j=1}^{n}\sigma_{ij} = n(n-1)\bar{\sigma_{ij}} \quad (i\ne j)$$

Substituting these definitions into the variance of an equally weighted portfolio gives:

$$\sigma_p^2= \dfrac{1}{n^2}n \bar{V} + \dfrac{1}{n^2}n(n-1)\bar{\sigma_{ij}} = \dfrac{1}{n}\bar{V} + \left( 1 - \frac{1}{n}\right) \bar{\sigma_{ij}}$$

As the number of assets $n$ tends to infinity, assuming average variance and average covariance remain well behaved:

$$\lim_{n \to \infty} \sigma_p^2 =  \bar{\sigma_{ij}} $$

**What remains is systematic (market-wide) risk that cannot be eliminated by diversification.**

Systematic risk affects many assets in the portfolio.
