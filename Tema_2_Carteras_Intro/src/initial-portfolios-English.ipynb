# FORMULAS USED IN PORTFOLIO ANALYSIS

In a risky situation, a rational investor will try to maximize the expected utility of the portfolio.

If the utility function is quadratic or returns follow a normal distribution:

utility is a function of the **mean** and the **variance** of portfolio returns.

___

#### LET US REVIEW THE FORMULAS FOR THE MEAN AND VARIANCE OF AN ASSET:

Let:

* $R_i$ be the return of asset $i$

* $R_{ij}$ be the possible returns of asset $i$

* $R_{ij}^*$ be the possible returns below the mean

* $P_i$ be the probability that a given return occurs

* $E(\tilde{R_i})$ be the expected return of asset $i$

Then:

* **Return** of an asset: $E(\tilde{R_i})=\sum_{j=1}^{m}R_{ij}P_j$

* **Variance** of an asset: $Var(\tilde{R_i})=\sigma_i^2=\sum_{j=1}^{m}(R_{ij}-E(\tilde{R_i}))^2P_j$

* **Standard deviation** of an asset: $\sigma_i=\sqrt{\sigma_i^2}=\sqrt{\sum_{j=1}^{m}(R_{ij}-E(\tilde{R_i}))^2P_j}$

* **Semivariance** of an asset: $SV_i=\sum_{j=1}^{m}(R_{ij}^*-E(\tilde{R_i}))^2P_j$  
  (only returns below the mean are taken into account)

___

#### LET US REVIEW THE FORMULAS FOR THE MEAN AND VARIANCE OF A TWO-ASSET PORTFOLIO:

Let:

* $w_1$ be the weight of asset 1 in the portfolio

* $w_2$ be the weight of asset 2 in the portfolio

* $E(\tilde{R_p})$ be the expected return of the portfolio

* $\sigma^2(\tilde R_p)$ be the variance of the portfolio

* $\rho_{12}$ be the correlation between assets 1 and 2

Then:

- The portfolio return is:
    - $E(\tilde R_P)=w_1E(\tilde R_1)+w_2E(\tilde R_2)$  
      Remember that portfolio return does not depend on the relationship between the assets.

- The portfolio variance is:
    - $\sigma^2(\tilde R_p)=\underset{Asset\ 1}{\underbrace{w_1^2\sigma^2(\tilde R_1)}}+\underset{Asset\ 2}{\underbrace{w_2^2\sigma^2(\tilde R_2)}}+\underset{Relationship\ between\ assets}{\underbrace{2w_1w_2\rho_{12}\sigma_1\sigma_2}}$

- Portfolio volatility is the square root of portfolio variance:
    - $\sigma(\tilde R_p)=\sqrt{\sigma^2(\tilde R_P)}=\sqrt{w_1^2\sigma^2(\tilde R_1)+w_2^2\sigma^2(\tilde R_2)+2w_1w_2\rho_{12}\sigma_1\sigma_2}$

___

#### LET US REVIEW THE FORMULAS FOR THE MEAN AND VARIANCE OF AN N-ASSET PORTFOLIO:

Let:

* $N$ be the number of assets in the portfolio

* $w_i$ be the proportion invested in asset $i$ $\Rightarrow \sum_{i=1}^{N}w_i=1$

* $\tilde R_p$ be the portfolio return $\Rightarrow \tilde R_p=\sum_{i=1}^{N}w_i\tilde R_i$

* $\sigma_{ij}$ be the covariance between assets $i$ and $j$

Then:

* **Portfolio return**: $E(\tilde R_p)=\sum_{i=1}^{N}w_iE(\tilde R_i)$

* **Portfolio variance**:  
  $\sigma^2(\tilde R_p)=\sigma_p^2=
  \underset{Asset\ variances}{\underbrace{\sum_{i=1}^{N}w_i^2}\sigma_i^2}
  +
  \underset{Dependence\ between\ assets}{\underbrace{\sum_{i=1}^{N}\sum_{j=1}^{N}w_iw_j\sigma_{ij}}}$ [¹]

* **Correlation coefficient** between assets $i$ and $j$:  
  $\rho_{ij}=\dfrac{\sigma_{ij}}{\sigma_i\sigma_j}$

[¹] $\sigma^2(\tilde R_p)=w_1^2\sigma^2(\tilde R_1)+w_2^2\sigma^2(\tilde R_2)+...+w_N^2\sigma^2(\tilde R_N)+2w_1w_2cov(\tilde R_1,\tilde R_2)+2w_1w_3cov(\tilde R_1,\tilde R_3)+...+2w_{N-1}w_Ncov(\tilde R_{N-1},\tilde R_N)$

The portfolio variance can be expressed in matrix form as:

$\sigma^2(\tilde R_p)=
\begin{pmatrix}w_1 & w_2 & ... & w_N\end{pmatrix}
\begin{pmatrix}
\sigma_1^2 & \sigma_{12}^2 & ... & \sigma_{1N}^2\\
\sigma_{21}^2 & \sigma_2^2 & ... & \sigma_{2N}^2\\
... & ... & ... & ...\\
\sigma_{N1}^2 & \sigma_{N2}^2 & ... & \sigma_N^2
\end{pmatrix}
\begin{pmatrix}w_1\\w_2\\...\\w_N\end{pmatrix}
=\mathbf{w}^T\mathbf{C}\mathbf{w}$

___

#### CONTRIBUTION OF AN ASSET TO PORTFOLIO RETURN AND RISK

Contribution of asset $i$ to:

* **Portfolio return**: $w_iE(\tilde R_i)$

* **Portfolio risk**: $w_i^2cov(\tilde R_i,\tilde R_P)=w_i^2\sigma_{i,P}$

**CONCLUSIONS:**

* The contribution of asset $i$ to portfolio return depends on its weight $w_i$, but not on the composition of the portfolio.

* The contribution of asset $i$ to portfolio risk depends on its weight $w_i$ and on the composition of the portfolio. The same asset may contribute more risk to some portfolios than to others.

For example, if an asset has a positive expected return and a negative covariance with the portfolio:

* it will **increase portfolio return**

* it will **reduce portfolio risk**.
