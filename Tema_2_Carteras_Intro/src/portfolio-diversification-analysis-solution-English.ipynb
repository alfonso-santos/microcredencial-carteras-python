## 1. The Effect of Diversification on Portfolio Risk

We have seen that the variance of a two-asset portfolio is given by:

$$\sigma^2_p= \underset{\text{Asset 1}} {\underbrace{w_1^2\sigma^2_1}}+\underset{\text{Asset 2}} {\underbrace{w_2^2\sigma^2_2}}+ \underset{\text{Relationship between assets}} {\underbrace{2w_1w_2 \sigma_{12}}}$$

We can generalize this expression to a portfolio with $n$ assets:

$$\sigma^2_p = \underset{\text{Individual asset risk}} {\underbrace{\sum_{i=1}^n w_i^2 \sigma_i^2}} + \underset{\text{Relationship between assets}} {\underbrace{ \sum_{i=1}^n \sum_{j=1}^n w_i w_j \sigma_{ij}}} \quad \forall i \neq j$$

If we construct an equally weighted portfolio, investing the same amount in each asset, $\Rightarrow w_i = \dfrac{1}{n}$, its variance can be written as:

$$\sigma_p^2= \dfrac{1}{n^2}\sum_{i=1}^{n} \sigma _i^2 + \dfrac{1}{n^2}\sum_{i=1}^{n}\sum_{j=1}^{n}\sigma_{ij} \quad \forall i \neq j$$

Define:

* **Average asset variance**

$$\bar{V} = \dfrac{1}{n}\sum_{i=1}^{n} \sigma _i^2 \quad \Rightarrow \quad \sum_{i=1}^{n} \sigma _i^2 = n \bar{V}$$

* **Average covariance between assets**

$$\bar{\sigma_{ij}} = \dfrac{1}{n(n-1)}\sum_{i=1}^{n}\sum_{j=1}^{n}\sigma_{ij} \quad \Rightarrow \quad  \sum_{i=1}^{n}\sum_{j=1}^{n}\sigma_{ij} = n(n-1)\bar{\sigma_{ij}}$$

Substituting into the variance equation for an equally weighted portfolio:

$$\sigma_p^2= \dfrac{1}{n^2}n \bar{V} + \dfrac{1}{n^2}n(n-1)\bar{\sigma_{ij}} = \dfrac{1}{n}\bar{V} + \left( 1 - \frac{1}{n}\right) \bar{\sigma_{ij}}$$

As the number of assets $n$ approaches infinity:

$$\lim_{n \to \infty} \sigma_p^2 =  \bar{\sigma_{ij}} $$

**What remains is systematic (market) risk, which cannot be eliminated through diversification.**

Systematic risk is the risk that affects all assets in a portfolio.

```python
# Load asset prices.
precios = pd.read_excel("..\data\SP500_comp_data.xlsx", index_col="Date")
tickers = precios.columns
```

```python
# Clean the data.
precios = precios.dropna(axis=1, how="all")
precios = precios.dropna(axis=0, how="any")
# Calculate log returns.
retornos = np.log(precios).diff().dropna()
```

```python
desviaciones = []

# Build equally weighted portfolios, starting with 2 assets and adding one asset at a time.
for num_activos in range(2, retornos.shape[1], 1):
    # Asset returns.
    rets = retornos.iloc[:, :num_activos]
    # Asset weights.
    pesos = np.ones(num_activos) / num_activos
    # Portfolio returns.
    retornos_cartera = rets @ pesos
    # Portfolio standard deviation.
    std = retornos_cartera.std()
    # Store the portfolio standard deviation.
    desviaciones.append(std)
```

```python
# Calculate the covariance matrix.
covarianzas = retornos.cov()
# Keep the lower triangular part of the matrix.
covarianzas = covarianzas.where(np.tril(np.ones(covarianzas.shape), k=-1).astype(bool))
# Stack the matrix.
covarianzas = covarianzas.stack().reset_index()
# Rename the columns.
covarianzas.columns = ["Asset 1", "Asset 2", "Covariance"]
covarianzas
```

Let us create a simple chart.

```python
plt.figure(figsize=(10, 4))
plt.title("Standard Deviation of Equally Weighted Portfolios")
plt.xlabel("Number of Assets")
plt.ylabel("Standard Deviation")
plt.plot(desviaciones, label="Standard Deviation")
std_mean = np.sqrt(covarianzas.loc[:, "Covariance"].mean())
plt.axhline(y=std_mean, color="r", linestyle="--", alpha=0.2, label="Average Standard Deviation")
plt.legend()
plt.show()
```

Let us create a slightly more detailed chart.

```python
# Create the figure.
plt.figure(figsize=(16, 4))

# Volatility chart.
plt.plot(desviaciones[0:50], label="Equally Weighted Portfolio Volatility")
plt.axhline(y=std_mean, color="r", linestyle="--", alpha=0.2, label="Average Volatility")
plt.xlabel("Number of Assets")
plt.ylabel("Volatility")
plt.title("Effect of Diversification on Risk", fontsize=16)
plt.legend()
plt.ylim(0, 0.025)

# Add an arrow.
plt.annotate(
    "",
    xy=(10, 0),
    xytext=(10, desviaciones[10]),
    arrowprops=dict(arrowstyle="<->", lw=1.5, color="black"),
)
plt.text(11, desviaciones[10] - 0.002, "Non-diversifiable Risk", fontsize=14, color="black")

plt.show()
```

## 3. Conclusion

1. The benefits of diversification have a limit: systematic risk, which cannot be eliminated.

2. Risk has two components: diversifiable risk (which can be reduced by adding assets) and systematic risk (which cannot be eliminated through diversification). Therefore, investors should not expect to be compensated for diversifiable risk, because it can be eliminated. Systematic risk is compensated, so we need to learn how to measure it and relate it to required returns.

3. With a relatively small number of assets, a portfolio can approach the level of systematic risk. Therefore, portfolios do not necessarily need a very large number of assets.

#### AN ASSET'S CONTRIBUTION TO PORTFOLIO RETURN AND RISK

Contribution of asset $i$ to:

* **Portfolio return:** $w_i E(\tilde R_i)$

* **Portfolio risk:** $w_i^2cov(\tilde R_i, \tilde R_P)=w_i^2\sigma_{i,P}$

**CONCLUSIONS:**

* An asset's contribution to portfolio return depends on its weight $w_i$, but not on the composition of the portfolio.

* An asset's contribution to portfolio risk depends on both its weight $w_i$ and the composition of the portfolio. The same asset may contribute more risk to some portfolios than to others.

For example, if an asset has a positive expected return and a negative covariance with the portfolio, it will:

* **Increase the portfolio's expected return.**

* **Reduce the portfolio's risk.**
